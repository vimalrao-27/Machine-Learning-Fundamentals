```python
import pandas as pd

data = [
    [101, "mca"],
    [102, "mba"],
    [103, "m.e"],
    [104, "m.a"],
    [105, "m.com"]
]

df = pd.DataFrame(data, columns=["student", "course"])

df.set_index("student", inplace=True)

print(df)
```
