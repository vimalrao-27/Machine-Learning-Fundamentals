```python
import pandas as pd
```

```python
df = pd.read_csv("customer_orders.csv")
```

```python
print("Original Dataset:")
print(df)
```

```python
# a. Replace missing CustomerName with 'Unknown'
df["CustomerName"] = df["CustomerName"].fillna("Unknown")

```

```python
# b. Replace missing Product with 'Unknown Product'
df["Product"] = df["Product"].fillna("Unknown Product")

```

```python
# c. Drop rows with missing Quantity
df = df.dropna(subset=["Quantity"])
```

```python
# d. Replace missing PricePerUnit with mean
df["PricePerUnit"] = df["PricePerUnit"].fillna(
    df["PricePerUnit"].mean()
)
```

```python
# Convert negative Quantity values into positive values
df["Quantity"] = df["Quantity"].abs()
```

```python
# Identify number of duplicate rows
duplicate_count = df.duplicated().sum()

print("\nNumber of duplicate rows:", duplicate_count)
```

```python
# Remove duplicate rows
df = df.drop_duplicates()
```

```python
# Convert Quantity to integer
df["Quantity"] = df["Quantity"].astype(int)
```

```python
# TotalPrice = Quantity * PricePerUnit
df["TotalPrice"] = df["Quantity"] * df["PricePerUnit"]
```

```python
print("\nCleaned Dataset:")
print(df)
```

```python
print("\nData Types:")
print(df.dtypes)
```

```python
# Save cleaned dataset
df.to_csv("customer_orders_cleaned.csv", index=False)
```
