# Lab 4: Data Visualization and Insight Generation Using Python

### Finance Analytics - Visualization Questions
1. Show the number of transactions in each category using a count plot.
2. Show the number of transactions for each payment mode using a count plot.
3. Show the number of transactions by transaction type (Debit and Credit) using a count plot.
4. Debit vs Credit Amount Comparison (Bar Plot)
5. Spending Pattern by Customer Segment (Bar Plot)
6. Payment Mode Usage Distribution (Pie Chart)
7. Boxplot of Transaction Amount by Category

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Finance_Analytics_Dataset.csv")
df.head()
```

```python
sns.countplot(x="Category", data=df)
plt.title("Number of transactions in each category")
plt.show()
```

```python
sns.countplot(x="PaymentMode", data=df)
plt.title("Number of transactions for each payment mode")
plt.show()
```

```python
sns.countplot(x="TransactionType", data=df)
plt.title("Number of transactions by transaction type")
plt.show()
```

```python
sns.barplot(x='TransactionType', y='Amount', data=df)
plt.title("Debit vs Credit Amount Comparison")
plt.ylabel("Transaction Amount")
plt.show()
```

```python
sns.barplot(x='CustomerSegment', y='Amount', data=df)
plt.title("Total Spending by Customer Segment")
plt.ylabel("Total Amount")
plt.show()
```

```python
payment_counts = df['PaymentMode'].value_counts()
plt.pie(payment_counts, labels=payment_counts.index, autopct='%1.1f%%')
plt.title("Payment Mode Usage Distribution")
plt.show()
```

```python
sns.boxplot(data=df, x='Category', y='Amount')
plt.xticks(rotation=45)
plt.title("Transaction Amount by Category")
plt.show()
```
