# Lab 7: Building a Decision Tree Classification Model Using Python

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, export_text, plot_tree
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
df = pd.read_csv("iris_data.csv")
df
```

```python
X = df[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]
Y = df['species']
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=42)
```

```python
dtree = DecisionTreeClassifier(criterion='gini', max_depth=3, random_state=42)
dtree.fit(X_train, Y_train)
```

```python
Y_pred = dtree.predict(X_test)
print("Accuracy:", accuracy_score(Y_test, Y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(Y_test, Y_pred))
```

```python
print(export_text(dtree, feature_names=list(X.columns)))
```

```python
plt.figure(figsize=(14, 8))
plot_tree(dtree, feature_names=X.columns, class_names=dtree.classes_, filled=True)
plt.title("Decision Tree (max_depth=3)")
plt.show()
```
